```python
import json
import pandas as pd
```

```python
from pathlib import Path

base_folder = Path("learningBase")
train_folder = base_folder / "train"
validation_folder = base_folder / "validation"
activation_folder = Path("activationBase")

train_folder.mkdir(parents=True, exist_ok=True)
validation_folder.mkdir(parents=True, exist_ok=True)
activation_folder.mkdir(parents=True, exist_ok=True)
```

```python
training_data = pd.read_json("train/ft_train.json")
test_data = pd.read_json("validation/ft_validation.json")
```

```python
RANDOM_STATE = 666

training_data.to_csv("learningBase/train/training_data.csv", index=False)
test_data.to_csv("learningBase/validation/test_data.csv", index=False)

activation_data = test_data.sample(n=1, random_state=RANDOM_STATE)
activation_data.to_csv("activationBase/activation_data.csv", index=False)
```
