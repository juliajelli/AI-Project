```python
import json, re, random
import pandas as pd
import numpy as np
```

```python
data = pd.read_json("my_export.json")
data
```

```python
RANDOM_STATE = 666
VALIDATION_SIZE = 0.2

rng = np.random.default_rng(RANDOM_STATE)
indices = rng.permutation(len(data))

validation_len = int(len(data) * VALIDATION_SIZE)
validation_idx = indices[:validation_len]
train_idx = indices[validation_len:]

train = data.iloc[train_idx].reset_index(drop=True)
validation = data.iloc[validation_idx].reset_index(drop=True)
```

```python
train.to_json("ft_train.json")
validation.to_json("ft_validation.json")
```
